# Separate the Text-Reuse passages per title-year

Currently the tr passages are all written together, without any information on the title or year.
This is not ideal for the solr ingestion, so ideally they would be split by title year.

To do this, all current passages would need to be loaded onto memory, and then groupedby/filtered by title and year, to write them back to s3 in the correct organization.

## 1. Read in the current passages data

```python
input_bucket = "s3://41-processed-data-staging/text-reuse/text-reuse_v1-0-0"
#output_bucket = "s3://41-processed-data-staging/textreuse/textreuse_passim_v1-0-0"
output_bucket = "s3://40-processed-data-sandbox/textreuse/textreuse_passim_v1-0-0"
input_passages_part = "passim_output_run_2/tr_passages/"
output_passages_part = "results/tr_passages"

input_passages_path = os.path.join(input_bucket, input_passages_part, '*.jsonl.bz2')
input_passages_path
```

```python
passages = passim_data = db.read_text(
    input_passages_path, storage_options=IMPRESSO_STORAGEOPT
).map(json.loads).persist() 

passages.take(2)
```

```python
for provider, title_list in KNOWN_JOURNALS_DICT.items():

    msg = f"\n------------ {provider} ------------\n"
    print(msg)
    #logger.info(msg)
    if provider == 'BNF':
        for title_idx, title in enumerate(title_list):
            #if title == "DLE":
            msg = f"\n---- Filtering passages for {title} ({title_idx+1}/{len(title_list)}) ----"
            print(msg)
            #logger.info(msg)
            passages_for_title = passages.filter(lambda x: title in x['ci_id']).persist()
            # group by year and remove years without passages
            passages_for_title_year = passages_for_title.groupby(lambda x: x['date'].split('-')[0]).filter(lambda x: len(x[1])!=0)

            #passages_for_title_year.



            break
        break
```

```python
def print_data_types(elem, step):
    if isinstance(elem, tuple):
        if elem[1]['ci_id'] == 'excelsior-1919-10-19-a-i0027':
            print(f"step: {step}")
            print(f"isinstance(elem, tuple) - elem: {elem}")
            print(f"type(elem[1]): {type(elem[1])}")
    elif isinstance(elem, dict):
        if elem['ci_id'] == 'excelsior-1919-10-19-a-i0027':
            print(f"step: {step}")
            print(f"isinstance(elem, dict) - elem: {elem}")
            print(f"type(elem): {type(elem)}")
    elif isinstance(elem, list):
        if elem[0]['ci_id'] == 'excelsior-1919-10-19-a-i0027':
            print(f"step: {step}")
            print(f"isinstance(elem, list) - elem[0]: {elem[0]}, len(elem): {len(elem)}")
            print(f"type(elem): {type(elem)}, type(elem[0]): {type(elem[0])}")
    elif isinstance(elem, str):
        if 'excelsior-1919-10-19-a-i0027' in elem:
            print(f"step: {step}")
            print(f"elem[:100]: {elem[:100]}")

    return elem
```

```python
for year, filename in years_to_filenames:
    print(f"year: {year}, s3_path: {filename}")
    out_files = (
        passages_for_title_year.filter(lambda x: x[0]== year)
        #.map(lambda x: print_data_types(x, "post-filter".upper()))
        .map(lambda x: x[1])
        .flatten()
        #.map(lambda x: print_data_types(x, "only contents".upper()))
        .repartition(1)
        #.map(lambda x: print_data_types(x, "post-repartition".upper()))
        #.map(lambda x: x[0])
        .map(json.dumps)
        #.map(lambda x: print_data_types(x, "post-dumps".upper()))
        .to_textfiles(filename,storage_options=IMPRESSO_STORAGEOPT)
    )#.compute()
```
